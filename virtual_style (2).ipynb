```python
# Display Image Function
def display_images(image_list, title="Images"):
    for image_path in image_list:
        print(f"Processing: {image_path}")

        # Check if image file exists
        if os.path.exists(image_path):
            img = cv2.imread(image_path)
            if img is not None:
                # Use cv2_imshow() instead of cv2.imshow()
                cv2_imshow(img)
                cv2.waitKey(0)  # Wait until a key is pressed
                cv2.destroyAllWindows()
            else:
                print(f"Failed to load image: {image_path}")
        else:
            print(f"Image not found: {image_path}")

# Example image list
image_list =[ "/content/hourglass.jpg", "/content/image 2.jpeg", "path/to/nonexistent.jpg"]#add example of body type image
display_images(image_list, title="Example Images")
```

```python
# Input User Details
def get_user_details():
    try:
        print("Enter your body measurements (in cm):")
        height = float(input("Height: "))
        chest = float(input("Chest: "))
        waist = float(input("Waist: "))
        hips = float(input("Hips: "))

        print("\nEnter your skin tone (light/medium/dark):")
        skin_tone = input().strip().lower()
        while skin_tone not in ["light", "medium", "dark"]:
            print("Invalid skin tone. Please enter 'light', 'medium', or 'dark'.")
            skin_tone = input().strip().lower()

        print("\nEnter the occasion type (casual/office/party/function):")
        occasion = input().strip().lower()
        while occasion not in ["casual", "office", "party", "function"]:
            print("Invalid occasion type. Please enter 'casual', 'office', 'party', or 'function'.")
            occasion = input().strip().lower()

        return height, chest, waist, hips, skin_tone, occasion
    except ValueError:
        print("Invalid input. Please try again.")
        return get_user_details()
```

```python
# Body Shape Classification
def classify_body_shape(chest, waist, hips):
    if chest > hips and waist < hips:
        return "Hourglass"
    elif hips > chest and waist < chest:
        return "Pear"
    elif waist > chest and hips < chest:
        return "Apple"
    else:
        return "Rectangle"
```
